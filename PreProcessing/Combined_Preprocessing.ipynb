```python

import os

#Running this cell can clean multiple battery cells
# Project root folder
PROJECT_ROOT = os.path.abspath("..")  

RAW_FOLDER = os.path.join(PROJECT_ROOT, "data", "raw")
PROCESSED_FOLDER = os.path.join(PROJECT_ROOT, "data", "processed")

os.makedirs(PROCESSED_FOLDER, exist_ok=True)  # ensure folder exists
num_of_cells = 60
keys = [f"B000{num}" for num in range(9)]
keys.extend([f"B00{num}" for num in range(10, num_of_cells)])

filepaths = [os.path.join(RAW_FOLDER, f"{key}.mat") for key in keys]
outpaths = [os.path.join(PROCESSED_FOLDER, f"{key}Cleaned.parquet") for key in keys]








for index, key in enumerate(keys):
    filepath = filepaths[index]
    outpath = outpaths[index]
    
    if os.path.exists(filepath) and not os.path.exists(outpath):
        print(f"Processing {filepath}")
        con.convert(filepath, outpath, key, key)

 
```
